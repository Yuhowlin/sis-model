```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from glob import glob

def read_and_process_csv(directory):
    """
    讀取指定資料夾內（包含子資料夾）的 CSV 檔案，並對相同實驗參數的數據取平均與標準差
    """
    # 遞迴搜尋所有 CSV 檔案 (包含子資料夾內的)
    csv_files = glob(os.path.join(directory, "**", "*.csv"), recursive=True)
    if not csv_files:
        print("❌ 未找到任何 CSV 檔案！請確認資料夾路徑是否正確")
        return
    
    # 儲存所有數據
    grouped_data = {}

    for file in csv_files:
        df = pd.read_csv(file)

        # 取得作為索引的參數 (去除步數 & 感染人數)
        param_keys = ["Infection Rate", "Recovery Rate", "Total People", "Initial Infected",
                      "Num Clusters", "Intra-cluster Connections", "Inter-cluster Connections",
                      "Total Steps", "Measure Interval", "Method"]

        param_values = tuple(df.iloc[0][param_keys])  # 取第一行當作識別這組實驗參數

        # 依據實驗參數存入字典
        if param_values not in grouped_data:
            grouped_data[param_values] = []
        
        grouped_data[param_values].append(df["Infected Count"] / df["Total People"])  # 存感染比例

    # 計算平均與標準差
    results = {}
    for params, data_list in grouped_data.items():
        data_array = np.array(data_list)  # 轉為 numpy 陣列
        mean_values = np.mean(data_array, axis=0)
        std_values = np.std(data_array, axis=0)
        results[params] = (mean_values, std_values, df["Step"])  # 加入步數資訊
    
    return results

def plot_results(results):
    """
    繪製感染人數比例的變化，顯示平均值與標準差
    """
    plt.figure(figsize=(10, 6))

    for params, (mean_values, std_values, steps) in results.items():
        infection_rate, recover_rate, total_people, initial_infected, num_clusters, Intra_cluster_Connections, Inter_cluster_Connections, Total_Steps, *_ = params
        # label = f"num_clusters={num_clusters}"
        # label = f"Intra_cluster={Intra_cluster_Connections}, Inter_cluster={Inter_cluster_Connections}"
        label = f"infection_rate={infection_rate}, recover_rate={recover_rate}"

        plt.plot(steps, mean_values, label=label)
        plt.fill_between(steps, mean_values - std_values, mean_values + std_values, alpha=0.2)

    plt.xlabel("Step")
    plt.ylabel("Infected Population Ratio")
    plt.title("Epidemic Spread: Infection Ratio Over Time")
    plt.legend()
    plt.grid(True)
    plt.show()


# 🏁 讀取 CSV 並繪製圖表
directory = r"C:\Users\ASUS\Documents\GitHub\sis-model\results\classical\diff_infection"  # 📁 主目錄 (包含多個子資料夾)
processed_data = read_and_process_csv(directory)
if processed_data:
    plot_results(processed_data)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from glob import glob
from mpl_toolkits.mplot3d import Axes3D
from scipy.interpolate import griddata

def read_and_aggregate_data(directory):
    """
    讀取指定資料夾內的 CSV 檔案，對相同 (感染率, 恢復率) 的最終感染比例取平均與標準差
    """
    csv_files = glob(os.path.join(directory, "*.csv"))
    if not csv_files:
        print("❌ 未找到任何 CSV 檔案！請確認資料夾路徑是否正確")
        return None

    data_dict = {}  # 用於存儲 (infection_rate, recovery_rate) -> [final_infected_ratios]

    for file in csv_files:
        df = pd.read_csv(file)

        # 取得實驗參數
        infection_rate = df.iloc[0]["Infection Rate"]
        recovery_rate = df.iloc[0]["Recovery Rate"]
        total_people = df.iloc[0]["Total People"]

        # 取得最終感染比例（最後一步的數據）
        final_infected_ratio = df["Infected Count"].iloc[-1] / total_people

        # 依照 (感染率, 恢復率) 分類
        key = (infection_rate, recovery_rate)
        if key not in data_dict:
            data_dict[key] = []
        data_dict[key].append(final_infected_ratio)

    # 計算平均與標準差
    aggregated_data = []
    for (infection_rate, recovery_rate), ratios in data_dict.items():
        avg_ratio = np.mean(ratios)
        std_ratio = np.std(ratios)
        aggregated_data.append((infection_rate, recovery_rate, avg_ratio, std_ratio))

    return np.array(aggregated_data)

def plot_3d_surface(data):
    """
    繪製 3D 平滑曲面圖：
    X = 感染率, Y = 恢復率, Z = 最終感染比例（取平均）
    """
    fig = plt.figure(figsize=(10, 7))
    ax = fig.add_subplot(111, projection='3d')

    # 提取 X, Y, Z 資料
    x = data[:, 0]  # 感染率
    y = data[:, 1]  # 恢復率
    z = data[:, 2]  # 最終感染比例（平均值）

    # 生成平滑的插值曲面
    xi = np.linspace(min(x), max(x), 30)
    yi = np.linspace(min(y), max(y), 30)
    xi, yi = np.meshgrid(xi, yi)
    zi = griddata((x, y), z, (xi, yi), method='cubic')

    # 繪製 3D 曲面
    surf = ax.plot_surface(xi, yi, zi, cmap='coolwarm', alpha=0.8, edgecolor='k')

    # 設定標籤
    ax.set_xlabel("Infection Rate")
    ax.set_ylabel("Recovery Rate")
    ax.set_zlabel("Final Infection Ratio")
    ax.set_title("Final Infection Ratio vs Infection & Recovery Rates (Averaged)")

    fig.colorbar(surf, shrink=0.5, aspect=10)  # 添加顏色條
    plt.show()

# 🏁 讀取 CSV 並繪製 3D 曲面圖
directory = r"C:\Users\ASUS\Documents\GitHub\sis-model\results\quantum"  # 📁 你的 CSV 存放目錄
data = read_and_aggregate_data(directory)
if data is not None:
    plot_3d_surface(data)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from glob import glob

def read_and_process_csv(directory):
    """
    讀取指定資料夾內（包含子資料夾）的 CSV 檔案，並對相同感染率的數據取平均與標準差
    """
    # 遞迴搜尋所有 CSV 檔案 (包含子資料夾內的)
    csv_files = glob(os.path.join(directory, "**", "*.csv"), recursive=True)
    if not csv_files:
        print("❌ 未找到任何 CSV 檔案！請確認資料夾路徑是否正確")
        return
    
    # 儲存所有數據
    grouped_data = {}

    for file in csv_files:
        df = pd.read_csv(file)

        # 取得感染率、恢復率
        infection_rate = df["Infection Rate"].iloc[0]
        recovery_rate = df["Recovery Rate"].iloc[0]

        # 取得最終感染比例 (最後一個 step 的 Infected Count / Total People)
        final_infected_ratio = df["Infected Count"].iloc[-1] / df["Total People"].iloc[-1]

        # 依據感染率分組
        if infection_rate not in grouped_data:
            grouped_data[infection_rate] = []
        
        grouped_data[infection_rate].append(final_infected_ratio)

    # 計算平均與標準差
    results = {}
    for infection_rate, data_list in grouped_data.items():
        mean_value = np.mean(data_list)
        std_value = np.std(data_list)
        results[infection_rate] = (mean_value, std_value)
    
    return results

def plot_results(results):
    """
    繪製最終感染比例與感染率的關係，並標記標準差
    """
    infection_rates = sorted(results.keys())  # 確保 x 軸有序
    mean_values = [results[rate][0] for rate in infection_rates]
    std_values = [results[rate][1] for rate in infection_rates]

    plt.figure(figsize=(10, 6))
    plt.plot(infection_rates, mean_values, marker="o", linestyle="-", label="Final Infected Ratio")
    plt.fill_between(infection_rates, 
                     np.array(mean_values) - np.array(std_values), 
                     np.array(mean_values) + np.array(std_values), 
                     alpha=0.2, label="Standard Deviation")

    plt.xlabel("Infection Rate")
    plt.ylabel("Final Infected Population Ratio")
    plt.title("Impact of Infection Rate on Final Epidemic Size")
    plt.legend()
    plt.grid(True)
    plt.show()

directory = r"C:\Users\ASUS\Documents\GitHub\sis-model\results\classical\diff_infection_scale-free"  # 📁 你的 CSV 存放目錄
processed_data = read_and_process_csv(directory)
if processed_data:
    plot_results(processed_data)
```

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from glob import glob

def read_and_process_csv(directory):
    """
    讀取指定資料夾內（包含子資料夾）的 CSV 檔案，並對相同感染率的數據取平均與標準差
    """
    # 遞迴搜尋所有 CSV 檔案 (包含子資料夾內的)
    csv_files = glob(os.path.join(directory, "**", "*.csv"), recursive=True)
    if not csv_files:
        print("❌ 未找到任何 CSV 檔案！請確認資料夾路徑是否正確")
        return
    
    # 儲存所有數據
    grouped_data = {}

    for file in csv_files:
        df = pd.read_csv(file)

        # 取得感染率、恢復率
        Measure_Interval = df["Measure Interval"].iloc[0]
        recovery_rate = df["Recovery Rate"].iloc[0]

        # 取得最終感染比例 (最後一個 step 的 Infected Count / Total People)
        final_infected_ratio = df["Infected Count"].iloc[-1] / df["Total People"].iloc[-1]

        # 依據感染率分組
        if Measure_Interval not in grouped_data:
            grouped_data[Measure_Interval] = []
        
        grouped_data[Measure_Interval].append(final_infected_ratio)

    # 計算平均與標準差
    results = {}
    for Measure_Interval, data_list in grouped_data.items():
        mean_value = np.mean(data_list)
        std_value = np.std(data_list)
        results[Measure_Interval] = (mean_value, std_value)
    
    return results

def plot_results(results):
    """
    繪製最終感染比例與感染率的關係，並標記標準差
    """
    Measure_Intervals = sorted(results.keys())  # 確保 x 軸有序
    mean_values = [results[rate][0] for rate in Measure_Intervals]
    std_values = [results[rate][1] for rate in Measure_Intervals]

    plt.figure(figsize=(10, 6))
    plt.plot(Measure_Intervals, mean_values, marker="o", linestyle="-", label="Final Infected Ratio")
    plt.fill_between(Measure_Intervals, 
                     np.array(mean_values) - np.array(std_values), 
                     np.array(mean_values) + np.array(std_values), 
                     alpha=0.2, label="Standard Deviation")

    plt.xlabel("Measure_Interval number")
    plt.ylabel("Final Infected Population Ratio")
    plt.title("Impact of Measure Interval on Final Epidemic Size")
    plt.legend()
    plt.grid(True)
    plt.show()


# 🏁 讀取 CSV 並繪製圖表
directory = r"C:\Users\ASUS\Documents\GitHub\sis-model\results\classical\diff_Measure_Interval_random"  # 📁 你的 CSV 存放目錄
processed_data = read_and_process_csv(directory)
if processed_data:
    plot_results(processed_data)
```

```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
import powerlaw

def analyze_network_properties(G, title="Network"):
    print(f"\n--- {title} ---")
    
    # 平均聚類係數
    clustering_coeff = nx.average_clustering(G)
    print(f"Average Clustering Coefficient: {clustering_coeff:.4f}")
    
    # 平均最短路徑長度（需保證圖是連通的）
    if nx.is_connected(G):
        avg_shortest_path = nx.average_shortest_path_length(G)
        print(f"Average Shortest Path Length: {avg_shortest_path:.4f}")
    else:
        print("Graph is not connected. Skipping shortest path length.")

    # 畫出度分布
    degrees = [deg for node, deg in G.degree()]
    plt.figure()
    degree_count = Counter(degrees)
    x, y = zip(*sorted(degree_count.items()))
    plt.plot(x, y, 'bo-')
    plt.xlabel("Degree")
    plt.ylabel("Count")
    plt.title(f"Degree Distribution of {title}")
    plt.show()

    # Fit power-law
    try:
        fit = powerlaw.Fit(degrees, verbose=False)
        print(f"Power-law alpha: {fit.power_law.alpha:.4f}")
        print(f"Power-law xmin: {fit.power_law.xmin}")
        R, p = fit.distribution_compare('power_law', 'exponential')
        print(f"Power law vs exponential R={R:.4f}, p={p:.4f}")
    except Exception as e:
        print(f"Power-law fit failed: {e}")
```

```python
# 建立 Small-World 圖
sim._initialize_small_world_network()
analyze_network_properties(sim.graph, title="Small-World Network")

# 建立 Scale-Free 圖
sim._initialize_scale_free_network()
analyze_network_properties(sim.graph, title="Scale-Free Network")
```
